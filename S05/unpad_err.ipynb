```python
from cryptography.hazmat.primitives import padding

BLOCK_SIZE = 128 # 16 bytes 

def pad(ptxt):
    padder = padding.PKCS7(BLOCK_SIZE).padder()
    padded = padder.update(ptxt) + padder.finalize()
    return padded

def unpad(padded):
    unpadder = padding.PKCS7(BLOCK_SIZE).unpadder()
    ptxt = unpadder.update(padded) + unpadder.finalize()
    return ptxt
```

```python
mens = b"abc"

p = pad(mens)

p_corrompido = p[:-1] + b"\x02" #Se o valor que adicionamos no fim fosse =x01 o programa já funcionava

print("Mensagem corrompida:\n", p_corrompido)

t = len(p_corrompido)

print("Tamanho da mensagem:\n",t)


unpad(p_corrompido)

#   Como podemos observar, apesar do tamanho da mensagem ser multiplo do tamanho do bloco(16),
#este continua a dar erro. Isto acontece porque a mensagem já não segue a regra do PKCS7. 
#Esta regra diz nos que o valor de cada byte preenchido é igual ao valor de bytes adicionados.
#Por exemplo se for necessário adicionar um byte, o valor adicionado é 0x01. 
#     Posto isto, temos aqui um exemplo de um caso onde o tamanho continua multiplo do tamanho do bloco,
#mas contínua a dar erro devido à alteração do último byte para um valor que não respeita a regra anteriormente citada.
```
